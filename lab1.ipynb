```python
# импорт модулей
from tensorflow import keras
import matplotlib.pyplot as pyplot
import numpy as np
import sklearn

import random
from keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
# Создаём массивы на 500 элементов
TrainMassX=np.zeros((500,2))
TrainMassY=np.zeros(500)
# Проверяем размерность
print(TrainMassX.shape)
```

```python
# Генерируем датасет
for i in range (500):
  x=random.randint(-100,100)
  TrainMassX[i][0]=x
  y=random.randint(-100,100)
  TrainMassX[i][1]=y
  true_y=2*x+3*y+7
  TrainMassY[i]=true_y
```

```python
# Строим модель нейронной сети
model = Sequential()

# Первый слой и первое значение - это размерность входных данных. В нашем случае 2.
model.add(Dense(2, activation='relu'))

# Здесь описываются внутренние слои
model.add(Dense(3999, activation='relu'))

# Выходной слой со значением, равным
model.add(Dense(1))

# Компилируем модель. Указываем алгоритм оптимизатора. В нашем случае Adam и функцию потерь - это то, что минимизирует нейронная сеть.
model.compile(optimizer='adam', loss='mse')
```

```python
# Обучение нейронной сети
# epochs - количество эпох
# batch_size - количество записей, которые попадают в один проход обучения - влияет на скорость обучения
# validation_split - значение определяет сколько процентов данных идет на тест (валидацию)
# verbose - подробность логирования процесса обучения
history = model.fit(TrainMassX, TrainMassY, epochs=100, batch_size=10, validation_split=0.05, verbose=2)
```

```python
# вывод информации об обучении
pyplot.plot(history.history['loss'])
pyplot.plot(history.history['val_loss'])
pyplot.title('Потери на этапах проверки и обучения')
pyplot.ylabel('Потери')
pyplot.xlabel('Эпохи')
pyplot.legend(['Потери на этапе обучения', 'Потери на этапе проверки'], loc='upper left')
pyplot.show()
```
